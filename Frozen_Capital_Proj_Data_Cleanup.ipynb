```python
#Daily_Demand sheet - Standardizing

import re

daily_demand = pd.read_excel(
    frozen_capital,
    sheet_name="Daily_Demand",
    dtype=str)

dd = daily_demand

#Dates
dd["Date"] = pd.to_datetime(
    dd["Date"],
    format="mixed")

dd["Date"] = dd["Date"].dt.strftime("%m/%d/%Y")

#SKU
def clean_sku(sku):
    digits = re.findall(r"\d+", str(sku))

    if digits:
        return f"SKU{int(digits[0]):04d}"

    return sku

dd["SKU"] = dd["SKU"].str.strip()
dd["SKU"] = dd["SKU"].apply(clean_sku)

#Warehouse
dd["Warehouse"] = dd["Warehouse"].str.upper()
dd["Warehouse"] = dd["Warehouse"].str.replace("-", "", regex=False)
dd["Warehouse"] = dd["Warehouse"].str.replace(" ", "", regex=False)

#Validation/Review

print(sorted(dd["Date"].unique()),
      sorted(dd["SKU"].unique()),
      sorted(dd["Warehouse"].unique()),
      sep="\n")
```

```python
#Inventory_Transaction - Standardizing

inventory_transactions = pd.read_excel(
    frozen_capital,
    sheet_name="Inventory_Transactions",
    dtype=str)

it = inventory_transactions

#Date
it["Date"] = pd.to_datetime(
    it["Date"],
    format="mixed")

it["Date"] = it["Date"].dt.strftime("%m/%d/%Y")

#SKU
import re

def clean_sku(sku):
    digits = re.findall(r"\d+", str(sku))

    if digits:
        return f"SKU{int(digits[0]):04d}"

    return sku

it["SKU"] = it["SKU"].str.strip()
it["SKU"] = it["SKU"].apply(clean_sku)

#Validation / Review
print(sorted(dd["Date"].unique()),
      sorted(it["SKU"].unique()),
      sorted(it["Transaction_Type"].unique()),
      sep="\n")
```

```python
#Supplier_Master - Standardizing
supplier_master = pd.read_excel(
    frozen_capital,
    sheet_name="Supplier_Master",
    dtype=str)

sm = supplier_master

#Supplier ID
sm["Supplier_ID"] = sm["Supplier_ID"].str.strip()
sm["Supplier_ID"] = sm["Supplier_ID"].str.upper()

#Country
sm["Country"] = sm["Country"].str.strip()

#Supplier ID
sm["Supplier_ID"] = sm["Supplier_ID"].str.strip()
sm["Supplier_ID"] = sm["Supplier_ID"].str.upper()

#Supplier Name
sm["Supplier_Name"] = sm["Supplier_Name"].str.strip()

#Supplier Rating
sm["Supplier_Rating"] = sm["Supplier_Rating"].str.strip()
sm["Supplier_Rating"] = sm["Supplier_Rating"].str.upper()

#Contract Type
sm["Contract_Type"] = sm["Contract_Type"].str.strip()

#Validation/Review

print(sorted(sm["Supplier_ID"].unique()),
      sorted(sm["Country"].unique()),
      sorted(sm["Contract_Type"].unique()),
      sorted(sm["Supplier_Rating"].unique()),
      sep="\n")
```

```python
#Warehouse_Master - Standardizing

import re

warehouse_master = pd.read_excel(
    frozen_capital,
    sheet_name="Warehouse_Master",
    dtype=str)

wm = warehouse_master

#Warehouse
wm["Warehouse"] = wm["Warehouse"].str.upper()
wm["Warehouse"] = wm["Warehouse"].str.strip()

#Validation/Review
print(sorted(wm["Warehouse"].unique()),
      sorted(wm["Capacity"].unique()),
      sep="\n")
```
